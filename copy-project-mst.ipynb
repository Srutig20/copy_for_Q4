Q1 a)

```python
import requests

base = "https://eutils.ncbi.nlm.nih.gov/entrez/eutils/efetch.fcgi"

#defining accession ids for all group members' genes
accession_ids = ["NM_001093771.3", "NC_012920.1", "NM_007294.4"]

# defining function to fetch accession ids 
def fetch_fasta(accession): #accession is just a placeholder name for the function argument
    """Download one nucleotide record as FASTA and save it."""
    params = {
        "db": "nucleotide", "id": accession, "rettype": "fasta", "retmode": "text"
    }
    r = requests.get(base, params=params)
    if r.status_code != 200:
        return None
    filename = accession + ".fasta"
    with open(filename, "w") as fh:
        fh.write(r.text)
    return filename

# Looping through all accessions_ids
for acc in accession_ids:
    saved_file = fetch_fasta(acc)
    print("Saved:", saved_file)
```

Q1 b) 

```python
base = "https://eutils.ncbi.nlm.nih.gov/entrez/eutils/efetch.fcgi"

#defining accession ids for all group members' genes
accession_ids = ["NM_001093771.3", "NC_012920.1", "NM_007294.4"]

# defining function to fetch gb files using accession ids 
def fetch_gb(gb): #gb is just a placeholder name for the function argument
    """Download one nucleotide record as FASTA and save it."""
    params = {
        "db": "nucleotide", "id": gb, "rettype": "gb", "retmode": "text"
    }
    r = requests.get(base, params=params)
    if r.status_code != 200:
        return None
    filename_gb = gb + ".gb"
    with open(filename_gb, "w") as fh:
        fh.write(r.text)
    return filename_gb

# Looping through all accessions_ids
for ac in accession_ids:
    saved_gb_file = fetch_gb(ac)
    print("Saved:", saved_gb_file)
```

To get accession ids, cds coordinates, protein id, organism, total sequence length

```python
from Bio import SeqIO

gb_records = []

for i in range(len(accession_ids)):
    acc = accession_ids[i]
    filename = acc + ".gb"
    record = SeqIO.read(filename, "gb")
    gb_records.append(record)

protein_ids = ["NP_001087240.1", "YP_003024029.1", "NP_009225.1"]
cds_coords = ["25 .. 1974", "7586 .. 8269", "114 .. 5705"]

for a in range(len(accession_ids)):
    print("\nAccession Id:", accession_ids[a], sep="\t")
    print("length of sequence " +  ": ", len(gb_records[a].seq) )
    print("organism: ", gb_records[a].annotations["organism"])
    print("Protein ID: ", protein_ids[a])
    print("CDS coordinates: ", cds_coords[a])
```

protein sequence:

```python
base = "https://eutils.ncbi.nlm.nih.gov/entrez/eutils/efetch.fcgi"

#defining accession ids for all group members' genes

# defining function to fetch gb files using accession ids 
def fetch_pro(pro): #pro is just a placeholder name for the function argument
    """Download one protein record as FASTA and save it."""
    params = {
        "db": "protein", "id": pro, "rettype": "fasta", "retmode": "text"
    }
    r = requests.get(base, params=params)
    if r.status_code != 200:
        return None
    filename_pro = pro + ".fasta"
    with open(filename_pro, "w") as kh:
        kh.write(r.text)
    return filename_pro

# Looping through all accessions_ids
for accc in protein_ids:
    saved_pro_file = fetch_pro(accc)
    print("Saved:", saved_pro_file)
```

now to compare the 2 files, we import the protein fasta files into python memory:

```python
with open("protein_TXNRD1.fasta", "r") as ss:
    NCBI_prot_TXNRD1 = ss.read()
    NCBI_prot_TXNRD1 = NCBI_prot_TXNRD1.replace("\n", "")
    
with open("protein_COX2.fasta", "r") as sf:
    NCBI_prot_COX2 = sf.read()
    NCBI_prot_COX2 = NCBI_prot_COX2.replace("\n", "")

with open("protein_BRCA1.fasta", "r") as sd:
    NCBI_prot_BRCA1 = sd.read()
    NCBI_prot_BRCA1 = NCBI_prot_BRCA1.replace("\n", "")
```

```python
print(NCBI_prot_TXNRD1, NCBI_prot_COX2, NCBI_prot_BRCA1, sep="\n \n")
```

Now to compute the differences:

```python
print("Sequences from NCBI and translated are same for TXNRD1:", NCBI_prot_TXNRD1 == protein_TXNRD1)
print("Sequences from NCBI and translated are same for COX2:", NCBI_prot_COX2 == protein_COX2)
print("Sequences from NCBI and translated are same for BRCA1:", NCBI_prot_BRCA1 == protein_BRCA1)
```

```python
differences = 0

for a, b in zip(protein_TXNRD1, NCBI_prot_TXNRD1):
    if a != b:
        differences += 1
print ("differences in TXNRD1 protein sequences:", differences)
```

```python
differences1 = 0

for a, b in zip(protein_COX2, NCBI_prot_COX2):
    if a != b:
        differences1 += 1
print ("differences in COX2 protein sequences:", differences1)
```

```python
differences2 = 0

for a, b in zip(protein_BRCA1, NCBI_prot_BRCA1):
    if a != b:
        differences2 += 1
print ("differences in BRCA1 protein sequences:", differences2)
```
